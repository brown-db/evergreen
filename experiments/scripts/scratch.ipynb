```python
%cd /Users/alelee/src/brown-db/evergreen

import os
os.environ["EVERGREEN_CACHE_DIR_ROOT"] = "~/.cache/evergreen"

from evergreen import SessionContext, prompt, count_if, col
from evergreen.model.config import CortexModelConfig
from experiments.common import DEFAULT_LANGUAGE_MODEL, EMBEDDING_MODEL, CONNECTION_NAME
```

```python
ctx = SessionContext.create_optimized()
ctx.register_model_config(
    CortexModelConfig(
        [DEFAULT_LANGUAGE_MODEL],
        EMBEDDING_MODEL,
        CONNECTION_NAME,
    )
)
```

```python
df = ctx.read_json("/Users/alelee/src/brown-db/evergreen/data/twitter_customer_support/airlines.jsonl", key=("dialog_id",))
df = df.map(prompt("Identify whether the {dialog} involves a customer complaint about a booking issue", bool).alias("has_booking_issue")).aggregate([count_if(col("has_booking_issue")).alias("booking_issue_count")], group_by=[col("company_id")]).with_rank(col("booking_issue_count"))
result = df.collect()
```
